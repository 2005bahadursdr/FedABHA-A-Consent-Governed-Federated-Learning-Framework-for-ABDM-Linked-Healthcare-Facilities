# Notebook 11: Executive Visualizations & System Dashboard

## Objective
Generate publication-quality visualizations summarizing system architecture gains, FL convergence, Non-IID distributions, and ABHA Zero-Trust Consent Governance decisions.



```python
import os
import sys
# Ensure project root is in sys.path for backend and scripts imports
root_path = os.path.abspath('..') if os.path.basename(os.getcwd()) == 'notebooks' else os.path.abspath('.')
if root_path not in sys.path:
    sys.path.insert(0, root_path)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Subplot 1: Non-IID Skew
hospitals = ['Hospital A (Cardiology)', 'Hospital B (Diabetes)', 'Hospital C (Nephrology)']
hd_prev = [65, 15, 20]
diab_prev = [20, 70, 25]
ht_prev = [40, 35, 60]

x = np.arange(len(hospitals))
width = 0.25

axes[0, 0].bar(x - width, hd_prev, width, label='Heart Disease', color='#e74c3c')
axes[0, 0].bar(x, diab_prev, width, label='Diabetes', color='#3498db')
axes[0, 0].bar(x + width, ht_prev, width, label='Hypertension', color='#2ecc71')
axes[0, 0].set_title("1. Non-IID Data Heterogeneity Across Nodes")
axes[0, 0].set_ylabel("Prevalence (%)")
axes[0, 0].set_xticks(x)
axes[0, 0].set_xticklabels(hospitals, rotation=10)
axes[0, 0].legend()

# Subplot 2: FL Convergence
rounds = list(range(1, 11))
fl_acc = [0.65, 0.72, 0.78, 0.82, 0.85, 0.87, 0.88, 0.89, 0.894, 0.895]
axes[0, 1].plot(rounds, fl_acc, marker='o', linewidth=2.5, color='#8e44ad', label='FedAvg Global Model')
axes[0, 1].axhline(y=0.912, color='r', linestyle='--', label='Centralized Ideal Limit')
axes[0, 1].set_title("2. Multi-Round FL Convergence Trajectory")
axes[0, 1].set_xlabel("FL Communication Round")
axes[0, 1].set_ylabel("Global Test Accuracy")
axes[0, 1].legend()

# Subplot 3: DP Tradeoff
eps = [0.5, 1.0, 2.0, 5.0, 10.0]
accs = [0.78, 0.846, 0.875, 0.888, 0.892]
axes[1, 0].plot(eps, accs, marker='s', linewidth=2.5, color='#d35400')
axes[1, 0].set_title("3. Differential Privacy (DP-SGD) Utility Frontier")
axes[1, 0].set_xlabel("Privacy Budget (Epsilon)")
axes[1, 0].set_ylabel("Accuracy")

# Subplot 4: Zero-Trust Consent Governance Breakdown
consent_categories = ['Active Granted', 'Revoked Policy', 'Expired Period', 'Purpose Mismatch']
record_counts = [1250, 180, 45, 25]
axes[1, 1].pie(record_counts, labels=consent_categories, autopct='%1.1f%%', colors=['#2ecc71', '#e74c3c', '#f1c40f', '#e67e22'])
axes[1, 1].set_title("4. ABHA Zero-Trust Consent Decision Breakdown")

plt.tight_layout()
plt.show()

```
